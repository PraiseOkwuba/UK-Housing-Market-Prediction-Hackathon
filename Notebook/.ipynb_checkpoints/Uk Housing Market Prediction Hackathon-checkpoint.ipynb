## Step 3: 🛠 Feature Engineering Plan

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
import numpy as np

# Copy clean dataset
data = df.copy()

# Log transform target
data['Sale_Price_GBP'] = np.log1p(data['Sale_Price_GBP'])

# Feature Engineering
data['Age'] = 2025 - data['Year_Built']
data['Price_per_Sqft'] = np.log1p(data['Sale_Price_GBP'] / data['Square_Footage'])  # log scale to reduce skew
data['BedBath_Interaction'] = data['Bedrooms'] * data['Bathrooms']

# Extract temporal features from Listing_Date
data['Listing_Date'] = pd.to_datetime(data['Listing_Date'])
data['Listing_Year'] = data['Listing_Date'].dt.year
data['Listing_Month'] = data['Listing_Date'].dt.month
data['Listing_Season'] = data['Listing_Month'] % 12 // 3  # 0=Winter,1=Spring,2=Summer,3=Autumn

# Drop unneeded columns
drop_cols = ['Property_ID', 'Listing_Date', 'Revenue_Activity', 'Revenue_GBP_Monthly']
data = data.drop(columns=drop_cols)

# Encode categorical variables
categorical_cols = data.select_dtypes(include=['object']).columns
for col in categorical_cols:
    le = LabelEncoder()
    data[col] = le.fit_transform(data[col])

# Split features/target
X = data.drop(columns=['Sale_Price_GBP'])
y = data['Sale_Price_GBP']

# Scale numerical features for Linear Regression
scaler = StandardScaler()
X_scaled = X.copy()
X_scaled[X.columns] = scaler.fit_transform(X)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

## Step 4: Modelling

```python
# Models
models = {
    "RandomForest": RandomForestRegressor(random_state=42, n_estimators=200, max_depth=10),
    "XGBoost": XGBRegressor(random_state=42, n_estimators=200, max_depth=6, learning_rate=0.1),
    "LinearRegression": LinearRegression()
}

# Train and evaluate
results = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    mae = mean_absolute_error(np.expm1(y_test), np.expm1(y_pred))
    rmse = np.sqrt(mean_squared_error(np.expm1(y_test), np.expm1(y_pred)))
    r2 = r2_score(np.expm1(y_test), np.expm1(y_pred))
    results[name] = {"MAE": mae, "RMSE": rmse, "R2": r2}

results

# Format the latest model results (from `results_feat`) exactly as requested.

def fmt_currency(x):
    return f"£{x:,.0f}"

def fmt_r2(x):
    return f"{x:.2f}"

order = ["RandomForest", "XGBoost", "LinearRegression"]
names = {"RandomForest": "Random Forest", "XGBoost": "XGBoost", "LinearRegression": "Linear Regression"}

lines = ["📊 Model Performance (Improved)", ""]
for key in order:
    res = results[key]
    lines.append(names[key])
    lines.append("")
    lines.append(f"MAE: {fmt_currency(res['MAE'])}")
    lines.append("")
    lines.append(f"RMSE: {fmt_currency(res['RMSE'])}")
    lines.append("")
    lines.append(f"R²: {fmt_r2(res['R2'])}")
    lines.append("")

print("\n".join(lines))
```

```python
import os
import time
import json
import numpy as np
import mlflow
import mlflow.sklearn
import mlflow.xgboost
from mlflow.tracking import MlflowClient
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from mlflow.models.signature import infer_signature

# ===== Tracking config =====
mlflow.set_experiment("UK_Housing_Price_Prediction")

# Optional: enable autologging (comment out if you prefer only manual logging)
mlflow.sklearn.autolog(log_models=False)   # we’ll log models explicitly to control signatures
mlflow.xgboost.autolog(log_models=False)

# ===== Helper: evaluate on ORIGINAL price scale =====
def evaluate_on_original_scale(y_true_log, y_pred_log):
    y_true = np.expm1(y_true_log)
    y_pred = np.expm1(y_pred_log)
    mae  = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(((y_true - y_pred)**2).mean())
    r2   = r2_score(y_true, y_pred)
    return {"mae_gbp": mae, "rmse_gbp": rmse, "r2": r2}

# ===== Train and log each model as a nested MLflow run =====
parent_run_name = f"housing_pipeline_{time.strftime('%Y%m%d_%H%M%S')}"
with mlflow.start_run(run_name=parent_run_name) as parent_run:

    # Log some run-level metadata
    mlflow.set_tags({
        "project": "uk_housing",
        "target": "Sale_Price_GBP",
        "target_transformation": "log1p",
        "data_rows": len(X_train) + len(X_test),
        "features": ",".join(list(X_train.columns))
    })

    # Save your train/test split indices or a small data sample as artifacts (optional)
    os.makedirs("artifacts", exist_ok=True)
    pd.DataFrame(X_test.head(20)).to_csv("artifacts/sample_X_test.csv", index=False)
    pd.DataFrame(np.expm1(y_test.head(20))).rename(columns={0:"Sale_Price_GBP"}).to_csv("artifacts/sample_y_test.csv", index=False)
    mlflow.log_artifacts("artifacts", artifact_path="samples")

    # Loop through models
    for key, model in models.items():
        run_name = {"RandomForest":"Random Forest",
                    "XGBoost":"XGBoost",
                    "LinearRegression":"Linear Regression"}[key]

        with mlflow.start_run(run_name=run_name, nested=True):
            # Log model params (works for sklearn/xgb)
            try:
                mlflow.log_params(model.get_params())
            except Exception:
                pass

            # Fit & predict on the log scale (your current setup)
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)

            # Metrics on the ORIGINAL £ scale
            metrics = evaluate_on_original_scale(y_test, y_pred)
            mlflow.log_metrics({
                "MAE_GBP":  metrics["mae_gbp"],
                "RMSE_GBP": metrics["rmse_gbp"],
                "R2":       metrics["r2"]
            })

            # Pretty, human-readable summary as an artifact
            summary_txt = (
                f"{run_name}\n"
                f"MAE: £{metrics['mae_gbp']:,.0f}\n"
                f"RMSE: £{metrics['rmse_gbp']:,.0f}\n"
                f"R²: {metrics['r2']:.2f}\n"
            )
            with open("model_summary.txt", "w") as f:
                f.write(summary_txt)
            mlflow.log_artifact("model_summary.txt", artifact_path="reports")

            # Infer model signature using sample input & predicted (on log scale)
            # (Signature helps downstream serving)
            try:
                sig = infer_signature(X_test, y_pred)
            except Exception:
                sig = None

            # Log the trained model
            if key == "XGBoost":
                mlflow.xgboost.log_model(model, artifact_path="model", signature=sig, input_example=X_test.head(5))
            else:
                mlflow.sklearn.log_model(model, artifact_path="model", signature=sig, input_example=X_test.head(5))

print("Finished logging models to MLflow.")
```
